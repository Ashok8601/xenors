## Log a parameter, metric and model artifact
A useful run records enough context to reproduce the result, not only the final score.


````python
import mlflow
with mlflow.start_run():
    mlflow.log_param('max_depth',4)
    mlflow.log_metric('f1',0.83)
    mlflow.log_text('training complete','notes.txt')
    print(mlflow.active_run().info.run_id)
````

**Expected/checkpoint**
```text
<run-id>
```


## Compare runs programmatically
Ranking runs is only useful when the validation split and metric definition are consistent.


````python
runs = mlflow.search_runs(order_by=['metrics.f1 DESC'])
print(runs[['run_id','params.max_depth','metrics.f1']].head())
````

**Expected/checkpoint**
```text
run_id ... max_depth ... f1
```
