## Compare parameter counts
Parameter count affects memory and deployment cost, not just accuracy.


````python
from torchvision.models import resnet18, vgg16
for m in [resnet18(),vgg16()]:
    print(type(m).__name__, sum(p.numel() for p in m.parameters()))
````

**Expected/checkpoint**
```text
ResNet ... ~11.7M
VGG ... ~138M
```
