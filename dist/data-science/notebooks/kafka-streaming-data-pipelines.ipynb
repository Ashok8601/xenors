## Create a compact JSON event
Stable event schemas make downstream consumers easier to evolve.


````python
import json, time
event = {'device_id':'M-17','temp_c':71.2,'ts':int(time.time())}
payload = json.dumps(event).encode()
print(payload.decode())
````

**Expected/checkpoint**
```text
{"device_id": "M-17", "temp_c": 71.2, "ts": ...}
```


## Make the consumer idempotent
At-least-once delivery means consumers should tolerate repeats.


````python
seen=set()
def process(event_id, value):
    if event_id in seen: return 'duplicate skipped'
    seen.add(event_id)
    return f'processed {value}'
print(process('e1',42)); print(process('e1',42))
````

**Expected/checkpoint**
```text
processed 42
duplicate skipped
```
