## Inspect data before transforming it

```python
import pandas as pd

df = pd.read_csv('practice.csv')
print(df.shape)
print(df.dtypes)
print(df.head(3).to_string(index=False))
```

## Check missing values explicitly
Expected checkpoint: Only columns with at least one missing value are listed.


```python
missing = df.isna().sum().sort_values(ascending=False)
print(missing[missing.gt(0)])
```

## Create a reproducible summary
Expected checkpoint: A compact numeric summary appears with one row per numeric column.


```python
summary = df.select_dtypes('number').agg(['count','mean','median','min','max']).T
print(summary.round(2))
```
