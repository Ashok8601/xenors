## Verify a join with explicit keys
A left join preserves every order even when the customer dimension is incomplete.


````python
orders = spark.createDataFrame([(1,10),(2,20)], ['customer_id','amount'])
customers = spark.createDataFrame([(1,'A'),(3,'C')], ['customer_id','name'])
orders.join(customers,'customer_id','left').show()
````

**Expected/checkpoint**
```text
+-----------+------+----+
|customer_id|amount|name|
+-----------+------+----+
|1          |10    |A   |
|2          |20    |NULL|
```


## Read the physical plan
Use explain() before guessing about Spark performance.


````python
q = orders.groupBy('customer_id').sum('amount')
q.explain(mode='formatted')
````

**Expected/checkpoint**
```text
Physical Plan
+- HashAggregate ...
```
