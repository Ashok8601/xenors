## Compare cosine similarity
High cosine similarity means similar direction, not necessarily identical semantics.


````python
import numpy as np
a=np.array([1.,0.,1.]); b=np.array([.8,.1,.9])
cos=a@b/(np.linalg.norm(a)*np.linalg.norm(b))
print(round(cos,3))
````

**Expected/checkpoint**
```text
~0.994
```


## Apply metadata filtering
Metadata filters can improve relevance before vector ranking.


````python
records=[('doc1','policy'),('doc2','manual'),('doc3','policy')]
print([r for r in records if r[1]=='policy'])
````

**Expected/checkpoint**
```text
[('doc1', 'policy'), ('doc3', 'policy')]
```
