## Format one instruction example
Consistent formatting matters more than adding decorative prompt text.


````python
sample={'instruction':'Summarize','input':'RAG retrieves documents.','output':'RAG grounds answers with retrieved context.'}
text=f"Instruction: {sample['instruction']}\nInput: {sample['input']}\nAnswer: {sample['output']}"
print(text)
````

**Expected/checkpoint**
```text
Instruction: Summarize
Input: RAG retrieves documents.
Answer: RAG grounds answers with retrieved context.
```
