````python
import pandas as pd
df = pd.read_csv('../datasets/sql-window-functions-data-analysis.csv')
df.head()
````

## Baseline example
SQL to run in SQLite/PostgreSQL after loading the CSV:
```sql
WITH ranked AS (
  SELECT o.*,
         ROW_NUMBER() OVER (PARTITION BY customer_id ORDER BY order_date DESC) AS rn
  FROM orders o
)
SELECT * FROM ranked WHERE rn = 1;
```
Checkpoint: Run the cell and inspect the printed values before changing anything.


## Confirm table grain
SQL to run in SQLite/PostgreSQL after loading the CSV:
```sql
SELECT COUNT(*) AS rows, COUNT(DISTINCT customer_id) AS customers
FROM practice_orders;
```
Checkpoint: rows | customers
30   | 12


## Inspect duplicate keys
SQL to run in SQLite/PostgreSQL after loading the CSV:
```sql
SELECT customer_id, COUNT(*) AS n
FROM practice_orders
GROUP BY customer_id
HAVING COUNT(*) > 1
ORDER BY n DESC;
```
Checkpoint: Customers with more than one row are listed before any join or aggregation.


## Validate a grouped result
SQL to run in SQLite/PostgreSQL after loading the CSV:
```sql
SELECT region, COUNT(*) AS orders, ROUND(SUM(amount),2) AS revenue
FROM practice_orders
GROUP BY region
ORDER BY revenue DESC;
```
Checkpoint: region | orders | revenue
West   | ...    | ...


## Compare detail and aggregate totals
SQL to run in SQLite/PostgreSQL after loading the CSV:
```sql
WITH x AS (
  SELECT amount FROM practice_orders WHERE amount IS NOT NULL
)
SELECT ROUND(SUM(amount),2) AS detail_total,
       ROUND((SELECT SUM(amount) FROM x),2) AS check_total
FROM x;
```
Checkpoint: detail_total and check_total should match exactly.
