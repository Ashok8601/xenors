## Baseline example
Expected checkpoint: Run the cell and inspect the printed values before changing anything.


```python
import pandas as pd
s = pd.Series([20, 25, 25, 30, 35, 80, 20000])
print('mean', s.mean())
print('median', s.median())
print('std', s.std())
print('IQR', s.quantile(.75) - s.quantile(.25))
```

## Summarize the sample
Expected checkpoint: n, mean and median are printed from the first numeric column.


```python
import pandas as pd
import numpy as np

df = pd.read_csv('practice.csv')
x = df.select_dtypes('number').iloc[:,0].dropna()
print('n=', len(x), 'mean=', round(x.mean(),2), 'median=', round(x.median(),2))
```

## Measure spread
Expected checkpoint: Standard deviation and IQR provide two different views of spread.


```python
q1, q3 = x.quantile([0.25,0.75])
iqr = q3 - q1
print('std=', round(x.std(ddof=1),2), 'IQR=', round(iqr,2))
```

## Bootstrap one statistic
Expected checkpoint: The three values are the bootstrap 2.5th, 50th and 97.5th percentiles.


```python
rng = np.random.default_rng(42)
boot = [rng.choice(x, size=len(x), replace=True).mean() for _ in range(1000)]
print(np.percentile(boot,[2.5,50,97.5]).round(2))
```

## Check sensitivity to an extreme value
Expected checkpoint: The mean usually moves more than the median when one extreme value is added.


```python
x2 = pd.concat([x, pd.Series([x.max()*5])], ignore_index=True)
print('mean change:', round(x2.mean()-x.mean(),2))
print('median change:', round(x2.median()-x.median(),2))
```
