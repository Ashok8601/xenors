## Inspect partitions before tuning
Partition count affects parallelism; tune it after measuring stage time and shuffle size.


````python
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
df = spark.range(0, 1_000_000).repartition(8)
print(df.rdd.getNumPartitions())
print(df.groupBy().count().collect()[0][0])
````

**Expected/checkpoint**
```text
8
1000000
```


## Compare cached and uncached actions
Cache only reused data; caching everything can waste executor memory.


````python
df2 = df.withColumn('bucket', (df.id % 10))
df2.cache()
df2.count()
print(df2.groupBy('bucket').count().orderBy('bucket').show(3))
````

**Expected/checkpoint**
```text
bucket | count
0      | 100000
1      | 100000
2      | 100000
```
