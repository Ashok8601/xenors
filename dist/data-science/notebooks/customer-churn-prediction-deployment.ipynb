## Use a stratified split
Stratification preserves class ratios in small imbalanced datasets.


````python
from sklearn.model_selection import train_test_split
X=list(range(100)); y=[0]*80+[1]*20
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,stratify=y,random_state=42)
print(sum(ytr),sum(yte))
````

**Expected/checkpoint**
```text
16 4
```


## Tune the probability threshold
Choose thresholds from business cost and validation metrics, not 0.5 by habit.


````python
import numpy as np
p=np.array([.2,.4,.6,.8]); y=np.array([0,1,1,1])
for t in [.3,.5,.7]:
    pred=(p>=t).astype(int); print(t,(pred==y).mean())
````

**Expected/checkpoint**
```text
0.3 0.75
0.5 0.75
0.7 0.50
```
