## Calculate population stability index
PSI is a monitoring signal, not a universal retraining rule.


````python
import numpy as np
def psi(expected, actual, bins=10):
    q=np.quantile(expected,np.linspace(0,1,bins+1)); q[0]-=1e-9; q[-1]+=1e-9
    e=np.histogram(expected,q)[0]/len(expected); a=np.histogram(actual,q)[0]/len(actual)
    e=np.clip(e,1e-6,None); a=np.clip(a,1e-6,None)
    return np.sum((a-e)*np.log(a/e))
print(round(psi(np.random.normal(0,1,5000), np.random.normal(.4,1,5000)),3))
````

**Expected/checkpoint**
```text
~0.15 to 0.25 depending on sample
```


## Separate feature drift from performance drift
Monitor both data and outcome quality when labels become available.


````python
signals={'age_psi':0.18,'auc_baseline':0.84,'auc_current':0.81}
print('feature_drift',signals['age_psi']>0.1)
print('performance_drop',signals['auc_baseline']-signals['auc_current']>0.02)
````

**Expected/checkpoint**
```text
feature_drift True
performance_drop True
```
