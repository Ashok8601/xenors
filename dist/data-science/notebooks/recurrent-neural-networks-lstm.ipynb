## Build sliding windows without leakage
Create train/validation splits by time before fitting scalers.


````python
import numpy as np
s=np.arange(10)
X=np.array([s[i:i+3] for i in range(len(s)-3)])
y=np.array([s[i+3] for i in range(len(s)-3)])
print(X[:2]); print(y[:2])
````

**Expected/checkpoint**
```text
[[0 1 2]
 [1 2 3]]
[3 4]
```


## Inspect LSTM output shapes
Shape checks catch many sequence-model bugs early.


````python
import torch, torch.nn as nn
lstm=nn.LSTM(input_size=1,hidden_size=8,batch_first=True)
out,(h,c)=lstm(torch.randn(4,12,1))
print(out.shape,h.shape)
````

**Expected/checkpoint**
```text
torch.Size([4, 12, 8]) torch.Size([1, 4, 8])
```
