## Create a small warehouse and table
Auto-suspend is an easy first control for warehouse cost.


````python
CREATE WAREHOUSE IF NOT EXISTS ds_wh WAREHOUSE_SIZE='XSMALL' AUTO_SUSPEND=60;
CREATE TABLE IF NOT EXISTS sales(order_id INT, amount NUMBER(10,2));
INSERT INTO sales VALUES (1,120.50),(2,75.00);
SELECT COUNT(*) AS rows, SUM(amount) AS revenue FROM sales;
````

**Expected/checkpoint**
```text
ROWS | REVENUE
2    | 195.50
```
