## Validate request data with Pydantic
Schema validation prevents malformed requests reaching the model.


````python
from pydantic import BaseModel, Field
class ChurnRequest(BaseModel):
    tenure: int = Field(ge=0)
    monthly_charges: float = Field(gt=0)
print(ChurnRequest(tenure=12, monthly_charges=79.5).model_dump())
````

**Expected/checkpoint**
```text
{'tenure': 12, 'monthly_charges': 79.5}
```


## Smoke-test a prediction endpoint
A deployment is not done until an external client can call it successfully.


````python
import requests
r = requests.post('http://localhost:8000/predict', json={'tenure':12,'monthly_charges':79.5})
print(r.status_code)
print(r.json())
````

**Expected/checkpoint**
```text
200
{"prediction": 0, "probability": 0.18}
```
