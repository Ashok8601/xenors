## Freeze a pretrained backbone
Start with frozen features on small datasets, then unfreeze gradually if validation improves.


````python
from torchvision.models import resnet18, ResNet18_Weights
model=resnet18(weights=ResNet18_Weights.DEFAULT)
for p in model.parameters(): p.requires_grad=False
print(sum(p.requires_grad for p in model.parameters()))
````

**Expected/checkpoint**
```text
0
```


## Replace the classifier head
The new head is trainable even when the backbone remains frozen.


````python
import torch.nn as nn
model.fc=nn.Linear(model.fc.in_features,3)
print(model.fc)
````

**Expected/checkpoint**
```text
Linear(in_features=512, out_features=3, bias=True)
```
