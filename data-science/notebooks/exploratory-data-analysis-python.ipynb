```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('houses.csv')
print(df.info())
print(df.describe(include='all'))
print(df.isna().mean().sort_values(ascending=False).head())
df.plot.scatter(x='area_sqft', y='price', alpha=.4)
plt.show()
```

## Inspect plotting columns
Expected checkpoint: Check that numeric axes are numeric and categories are not accidentally encoded as measures.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('practice.csv')
print(df.dtypes)
print(df.head(3).to_string(index=False))
```

## Plot a distribution
Expected checkpoint: A histogram appears; look for skew, gaps and extreme values before summarizing the center.


```python
num = df.select_dtypes('number').columns[0]
ax = df[num].plot(kind='hist', bins=12, edgecolor='black')
ax.set(title=f'Distribution of {num}', xlabel=num)
plt.tight_layout()
```

## Compare groups
Expected checkpoint: The table shows group sizes beside mean and median so small groups are visible.


```python
cat = df.select_dtypes(exclude='number').columns[0]
num = df.select_dtypes('number').columns[0]
print(df.groupby(cat)[num].agg(['count','mean','median']).round(2))
```

## Save a reproducible figure
Expected checkpoint: practice-chart.png is saved with fixed dimensions and resolution.


```python
fig, ax = plt.subplots(figsize=(8,4.5))
df[num].plot(kind='box', ax=ax, vert=False)
ax.set_title(f'{num}: spread and possible outliers')
fig.tight_layout()
fig.savefig('practice-chart.png', dpi=160)
```
