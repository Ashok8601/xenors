## Validate model JSON instead of trusting it
Treat model output as untrusted input and validate it like an API response.


````python
import json
raw='{"priority":"high","score":0.87}'
obj=json.loads(raw)
assert obj['priority'] in {'low','medium','high'}
assert 0<=obj['score']<=1
print(obj)
````

**Expected/checkpoint**
```text
{'priority': 'high', 'score': 0.87}
```


## Create a tiny regression test set
Prompt changes should be tested against stable examples before deployment.


````python
tests=[('2+2','4'),('capital of France','Paris')]
for prompt,expected in tests:
    print(prompt,'-> expected:',expected)
````

**Expected/checkpoint**
```text
2+2 -> expected: 4
capital of France -> expected: Paris
```
