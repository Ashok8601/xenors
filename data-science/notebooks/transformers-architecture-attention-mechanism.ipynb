## Calculate scaled dot-product attention
Scaling prevents dot products from growing too large as key dimension increases.


````python
import torch, math
Q=torch.tensor([[1.,0.],[0.,1.]])
K=Q.clone(); V=torch.tensor([[10.,0.],[0.,20.]])
W=torch.softmax(Q@K.T/math.sqrt(2),dim=-1)
print(W.round(decimals=3)); print((W@V).round(decimals=2))
````

**Expected/checkpoint**
```text
tensor([[0.670, 0.330], [0.330, 0.670]])
...
```


## Verify causal masking
A causal mask prevents a decoder token from reading future tokens.


````python
scores=torch.zeros(4,4)
mask=torch.triu(torch.ones(4,4),diagonal=1).bool()
scores=scores.masked_fill(mask,float('-inf'))
print(scores)
````

**Expected/checkpoint**
```text
upper triangle becomes -inf
```
