## Compose a small runnable chain
Use small composable steps so you can test each stage independently.


````python
from langchain_core.runnables import RunnableLambda
clean=RunnableLambda(lambda x:x.strip().lower())
length=RunnableLambda(lambda x:{'text':x,'chars':len(x)})
print((clean|length).invoke('  Hello LangChain  '))
````

**Expected/checkpoint**
```text
{'text': 'hello langchain', 'chars': 15}
```


## Return structured data
Structured output is easier to validate than free-form text.


````python
from pydantic import BaseModel
class Answer(BaseModel):
    label:str
    confidence:float
print(Answer(label='positive',confidence=.91).model_dump())
````

**Expected/checkpoint**
```text
{'label': 'positive', 'confidence': 0.91}
```
