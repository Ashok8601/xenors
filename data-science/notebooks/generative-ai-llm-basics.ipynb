## Baseline example
Expected checkpoint: Run the cell and inspect the printed values before changing anything.


```python
prompt = '''You are given three customer notes.
Return JSON with: issue, urgency, next_action.
Do not invent facts that are not present.'''
print(prompt)
```

## Check tensor shapes first
Expected checkpoint: X: (32, 8) y: (32,)


```python
import numpy as np
X = np.random.default_rng(42).normal(size=(32, 8)).astype('float32')
y = (X[:,0] + X[:,1] > 0).astype('float32')
print('X:', X.shape, 'y:', y.shape)
```

## Define a tiny neural baseline
Expected checkpoint: The summary lists layer output shapes and parameter counts before training.


```python
import tensorflow as tf
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(8,)),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.summary()
```

## Train with validation data
Expected checkpoint: Final training and validation metrics are printed side by side.


```python
history=model.fit(X,y,epochs=5,batch_size=8,validation_split=.25,verbose=0)
print({k: round(v[-1],3) for k,v in history.history.items()})
```

## Inspect prediction confidence
Expected checkpoint: Probabilities are shown before thresholding, making uncertain predictions visible.


```python
p=model.predict(X[:5],verbose=0).ravel()
print(np.round(p,3))
print('labels:', (p>=.5).astype(int))
```
