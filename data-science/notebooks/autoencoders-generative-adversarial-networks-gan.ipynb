## Measure reconstruction error
Reconstruction loss is useful for autoencoders but does not directly measure perceptual quality.


````python
import torch
x=torch.tensor([[0.,1.],[1.,0.]])
x_hat=torch.tensor([[.1,.8],[.9,.2]])
print(torch.mean((x-x_hat)**2).item())
````

**Expected/checkpoint**
```text
0.025
```


## Track generator and discriminator losses together
GAN losses can look stable while samples collapse, so inspect generated outputs too.


````python
history={'g':[1.2,1.0,.9],'d':[.7,.65,.62]}
print('g_last',history['g'][-1],'d_last',history['d'][-1])
````

**Expected/checkpoint**
```text
g_last 0.9 d_last 0.62
```
