## Measure retrieval hit rate
Evaluate retrieval separately from answer generation so you know where failures originate.


````python
tests=[('q1','docA',['docA','docC']),('q2','docB',['docD','docB'])]
hits=sum(expected in retrieved for _,expected,retrieved in tests)
print('hit_rate@2 =',hits/len(tests))
````

**Expected/checkpoint**
```text
hit_rate@2 = 1.0
```


## Add citations to the assembled prompt
Preserving source IDs makes citations possible downstream.


````python
chunks=[('policy.pdf','Refunds are allowed within 30 days.'),('faq.md','Digital goods are final sale.')]
ctx='\n'.join(f'[{s}] {t}' for s,t in chunks)
print(ctx)
````

**Expected/checkpoint**
```text
[policy.pdf] Refunds are allowed within 30 days.
[faq.md] Digital goods are final sale.
```
