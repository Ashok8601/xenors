## Use walk-forward splits
Time-series validation must preserve chronology.


````python
series=list(range(12))
for cut in [6,8,10]:
    train=series[:cut]; test=series[cut:cut+2]
    print(len(train),test)
````

**Expected/checkpoint**
```text
6 [6, 7]
8 [8, 9]
10 [10, 11]
```


## Compare against a naive baseline
Complex models should beat simple persistence or seasonal baselines.


````python
y=[101,103,102,106]
pred=[100,101,103,102]
mae=sum(abs(a-b) for a,b in zip(y,pred))/len(y)
print('MAE',mae)
````

**Expected/checkpoint**
```text
MAE 2.0
```
