## Cache document fingerprints
Fingerprints help avoid re-embedding unchanged documents.


````python
import hashlib
text='example policy text'
print(hashlib.sha256(text.encode()).hexdigest()[:12])
````

**Expected/checkpoint**
```text
<12-char fingerprint>
```
