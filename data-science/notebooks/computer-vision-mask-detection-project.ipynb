## Check class balance before training
Class imbalance can make headline accuracy misleading.


````python
from collections import Counter
labels=['mask']*80+['no_mask']*20
print(Counter(labels))
````

**Expected/checkpoint**
```text
Counter({'mask': 80, 'no_mask': 20})
```


## Measure per-class recall
For safety-related detection, inspect the class whose misses matter most.


````python
from sklearn.metrics import classification_report
y_true=[0,0,0,1,1]; y_pred=[0,0,1,1,0]
print(classification_report(y_true,y_pred,digits=2))
````
