## Create rolling sensor features
Rolling features capture trend but must use only past data at prediction time.


````python
import pandas as pd
s=pd.Series([40,42,41,50,55])
print(pd.DataFrame({'temp':s,'roll3':s.rolling(3).mean()}))
````

**Expected/checkpoint**
```text
temp roll3
40 NaN
42 NaN
41 41.0
50 44.3
55 48.7
```


## Translate probabilities into maintenance priorities
Ranking risk can be more useful operationally than a hard binary prediction.


````python
assets=[('A',.12),('B',.81),('C',.46)]
print(sorted(assets,key=lambda x:x[1],reverse=True))
````

**Expected/checkpoint**
```text
[('B', 0.81), ('C', 0.46), ('A', 0.12)]
```
